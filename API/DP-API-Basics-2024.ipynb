#### Example 1: Exchange Rates API

```python
# API documentation: https://exchangeratesapi.io/documentation/
# Older version when you did not need authentication

import requests
import json

url = 'http://api.exchangeratesapi.io/latest'
exrate = requests.get(url)
jstr = exrate.text
print(jstr)
```

```python
# Starting 2021, need to obtain API key to access endpoints (free version): https://manage.exchangeratesapi.io/signup/free
# API documentation: https://exchangeratesapi.io/documentation/
# Free version provides access to two ways of extracting data (Latest, Historical)
```

```python
# print(jstr)
jp = exrate.json()
print(jp)
# data type - jp, jstr?
```

```python
for key in jp:
    print(key)
```

```python
# print the value of the key 'rates'
# USD
```

```python
# Cannot specify base currency to USD with the Free version of the API; but, we can call subset of rates

import api_keys
import requests
import json

PARAMETERS = {
                'access_key': api_keys.get_exchangerate_api(),
                'base': 'EUR',
                'symbols': 'USD,CAD,GBP'
             }

url = 'http://api.exchangeratesapi.io/v1/latest'
exrate = requests.get(url, params=PARAMETERS)
print(exrate.text)
```

```python
# Another way to access the json structure is to directly use exrate.json() rather than apply load string (loads) method on exrate.text
# https://www.geeksforgeeks.org/json-loads-in-python/

exrate_jp = json.loads(exrate.text)
print(type(exrate_jp))
```

```python
jp=exrate.json()
print(type(jp))
```

```python
for x,y in exrate_jp['rates'].items():
    print (x,y)
```

```python
for x,y in jp['rates'].items():
     print (x,y)
```

##### Endpoint #2 : Obtaining Historical Exchange Rates

```python

# Historical Data

import api_keys
import requests
import json

PARAMETERS = {
                'access_key': api_keys.get_exchangerate_api(),
                'base': 'EUR',
                'symbols': 'USD,CAD,GBP'
             }

url_hist = 'http://api.exchangeratesapi.io/v1/2020-12-30'
exrate = requests.get(url_hist, params=PARAMETERS)
print(exrate.text)
```
