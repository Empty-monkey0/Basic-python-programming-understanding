#### Syntax Errors, Runtime Errors and Logic Errors

A syntax error occurs when the language coding rules are violated.  For example, we would encounter a syntax error if a colon did not terminate a def main() statement.

A runtime error occurs when a problem that cannot be detected until the code runs is encountered when the program runs; this is illustrated in the cell below.

A logic error is something that isn't caught, either when Python is interpreting the code or at runtime, but which causes an issue. For example, if you intended to multiply two numbers together, but instead added them, that would be a logic error.

```python
# Definition of the main function.
def main():
    get_name()
    print(f'Hello {name}.')     # This causes an error!
                                # name is local to get_name()

# Definition of the get_name function.
def get_name():
    name = input('Enter your name: ')

# Call the main function.
main()
```

##### The None Keyword

```python
# This program demonstrates the None keyword.

def main():
    # Get two numbers from the user.
    num1 = int(input('Enter a number: '))
    num2 = int(input('Enter another number: '))
    
    # Call the divide function.
    quotient = divide(num1, num2)

    # Display the result.
    if quotient is None:
        print('Cannot divide by zero.')
    else:
        print(f'{num1} divided by {num2} is {quotient}.')

# The divide function divides num1 by num2 and
# returns the result. If num2 is 0, the function
# returns None.

def divide(num1, num2):
    if num2 == 0:
        result = None
    else:
        result = num1 / num2
    return result

# Execute the main function.
main()
```

##### Conditionally Executing the main() Function

A module usually contains functions.

But, can be run as a standalone program or can be imported.

When Python is processing a source code file, it creates a special variable named __ __name__ __.

We call this a "dunder" - a double underscore.

If the file is being imported as a module, the name variable will be set to the module name.

If the file is being executed as a standalone program, the name variable will be set to __main__.
 
Use the value of the __name__ variable to determine whether the main() function should execute.



```python
import random     # about two dozen methods
def main():
    for count in range(5):     #Sneak peek at next topic - loops 
        # Get a random number.
        number = random.randint(1, 100)
        
        # Display the number.
        print(f'The number is: {number}.')

# Call the main function.
if __name__ == '__main__':     # If the random module has a main(), it would be named random.
    main()                     # We want to run this program's main(), not random's main())
```
