- Variables whose values are declared as either True or False... or Boolean expressions that evaluate to either True or False

```python
a = True
b = False
c = 4==4           # assignment operator and Boolean expression with an equality operator
d = 4==5 or 4!=6   # OR logical operator to create compound Boolean expression
print ('a: ',a)
print ('b: ',b)
print ('c: ',c)
print ('d: ',d)

string1 = 'Hello' 

string2 = ''   # the empty string

# Built-in bool() function returns True if its argument is truthy and False if it is falsy.

print ('The Boolean value of a is',bool (a))

e = 1 
print ('The Boolean value of e is',bool (e))

f = 0

print ('The Boolean value of f is',bool (f))

g = -1

print ('The Boolean value of g is',bool (g))

h = 2

print ('The Boolean value of h is',bool (h))

print  ('The Boolean value of False is',bool (False))

print ('The Boolean value of string1 is',bool (string1))

print ('The Boolean value of string2 is',bool (string2))
```
