```python
# Program 3-4 (sort_names.py)
# This program compares strings with the < relational operator
# and demonstrates if-else.
# Get two names from the user.

name1 = input('Enter a name a name (last name first): ')
name2 = input('Enter another name (last name first): ')
    
# Display the names in alphabetical order.
print('Here are the names, listed alphabetically:')

if name1 < name2:   # header begins with keyword and ends with a colon; if clause or header
    print(name1)    # indented block or suite
    print(name2)    # indented block or suite
                    # the combination of the if statement (header) and the indented block (suite) form a clause
else:               # header; else clause or header; dual alternative decision structure
    print(name2)    # indented block or suite
    print(name1)    # indented block or suite
                    # The combination of if and else clauses forms a compound statement
```

```python
# This program demonstrates the multi-alternative decision structure (if-elif-else statement)

num_pass = int(input("Enter the number of passengers: "))
if num_pass <= 4: 
    print ('Choose a compact car')
elif num_pass == 5: 
    print ('Choose an intermediate car') 
elif num_pass == 6:
    print ('Choose a regular SUV') 
elif num_pass == 7: 
    print ('Choose a seven-passenger SUV') 
else:
    print ('Choose a bus') 

# Much easier to read    
```

## Ternary Operators
Ternary operators (also known as conditional expressions) test if a condition is true or false, and, depending on the outcome, returns the corresponding value.<br>  
syntax: [on_true] if [expression] else [on_false]

A compact alternative to the multiline if-else control flow statement in situations where we only need to switch between two values.

Ternary means there are three operands (each operand is an expression, not a statement):
- the value that will be returned if the condition is evaluated to be true
- a Boolean expression to test for true or false
- the value that will be returned if the condition is evaluated to be false

```python
# Program to demonstrate conditional operator
a, b = 10, 20    # multiple assignment
 
# Copy value of a to min if a < b else copy b to min
min = a if a < b else b   # Another example of syntactic sugar - an alternative to if ... else 
 
print (min)
```

```python
# Above program equivalent to this 

if a < b:
    min = a
else:
    min = b
    
print (min)
```
