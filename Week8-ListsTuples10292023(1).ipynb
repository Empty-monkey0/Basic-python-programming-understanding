# Copying lists
--- To make a copy of a list, you must copy the list’s elements.... WHY? In Python, assigning one variable to another variable simply makes both variables reference the same object in memory - see §7.6 (slide 28).

```python
#What will the following code print?
list1 = [3, 4]

print ('List 1 address at line 3:',id(list1))     # an object's id is its memory address
list2 = list1.copy()  # creates a new list

print ('List 2 address at line 5:',id(list2)) # Note different address from list 1
list1[0] = 99

list3 = list2    # list2 and list3 are both pointing to the same list (Chapter 7 slide 28) 
print ('List 3 address at line 8:',id(list3))

list3 +=list2    # Concatenated values, but both 2 and 3 still pointing to same list
print ('List 3 address at line 10:',id(list3))
print('List1:',list1, 'List2:',list2,'List3:', list3)

list4 = list3[:]   # Because a span is specified, a new list is created; compare to line 7 
print('List3:',id(list3), 'List4:',id(list4))

list4[3] = 59
print('List3:',list3, 'List4:',list4)  #Note that updating list4 did not affect list3
```

# A comparison of execution times using a loop and a list comprehension

```python
lst = [n for n in range(100000)]   # (Global) list of 100,000 items
# built using a list comprehension
 
def loops():      # A loop
    newlst = [] 
    for n in lst: 
        newlst.append(n * 2)      # Square the list element   
    return newlst 
 
def lstcomp():      # A list comprehension 
    return [n * 2 for n in lst]   # Square the list element
 
from timeit import timeit 
 
print(f' Time to process 1 million iterations using a loop {timeit(loops, number = 10)}') # Repeat loop function 10X 
 
print(f' Time to process 1 million iterations using a list comprehension {timeit(lstcomp, number = 10)}') # Repeat list comprehension 10X 
```

# Why did the list comprehension take ~ half of the time of the loop function?

Think of the nature of Python - every statement has to be interpreted.

The loops() function uses the append() function, while the lstcomp() function does not use a function - the overhead of repeatedly invoking append() is probably the cause of the higher execution time.

So, the first approach requires that Python loads and runs the .append() function repeatedly, while the list comprehension builds the entire list in one function call.`  

- __[timeit provides a simple way to time small bits of Python code.](https://https://docs.python.org/3/library/timeit.html)__

# Tuples -- almost like lists but immutable just as strings are immutable
- Then why Tuples? Faster performance and safer when you don't want to change the data, but just read in the data
- You can easily convert between list and tuples using list() and tuple() functions
- Tuples support all the same operations as lists, except those that change the contents of the list. Tuples do not support methods such as append(), remove(), insert(), reverse(), and sort().<br>

Dictionary definition: a tuple is a finite ordered list (sequence) of elements. An n-tuple is a sequence (or ordered list) of n elements, where n is a non-negative integer.
