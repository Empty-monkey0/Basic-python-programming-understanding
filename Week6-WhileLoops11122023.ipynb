## While loop program parts 

### 0. Setting up an initialization variable to be tested in the while loop's condition -- why? because while is a pretest loop!
### 1. A condition that is tested for a true or false value (similar to the if statement, it uses a Boolean expression to control the flow of execution)
>###     The condition should not simply be the keyword True; it should be an expression to be evaluated in the While statement; this way, you do not have to search through the code block to find the logic that controls the loop; it is at the top of the loop  
### 2. An indented statement or set of indented statements (block) that is ***repeated*** as long as the Boolean expression evaluates to true, and 
### 3. Some sort of indented statement within the while loop that will eventually cause the while condition to become false so that the loop is terminated! Otherwise, your code will keep executing in perpetuity (an infinite loop)!


initialization <br>
while continuationCondition : <br>
    -- do main action to be repeated <br>
    -- prepare variables for the next time through the loop <br>

```python
# Predict the output

temperature = 115  
while temperature > 112: # first while loop code
    print(temperature)
    temperature = temperature - 1

print('The tea is cool enough.')
```

```python
# A similar example of what not to do!


n = 1
while True:
    print (n)
    n = n + 1
    if n > 5:
        break     # Exit the loop
print ('done')
# what happens if I remove the if condition? If you try it, you will need to stop the kernel
# simpler way of writing the same code above!
```

## Sentinel values

#### IMPORTANT: Designate a carefully chosen value to indicate end of sequence of values.  For example, 0 for employee id.

```python
emp_ID = 1     # What would happen if this value was not initialized? Restart & clear output & try it!

while emp_ID != 0:         # 0 is a sentinel value
    emp_ID = int(input("Enter Employee ID:")) 
    print ('Employee ID is',emp_ID)
print ("Thank you")
```

# Augmented assignment operators
- A form of "syntactic sugar" - syntax within a programming language that is designed to make things easier to read or to express.
