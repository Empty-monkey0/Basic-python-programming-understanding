# Rules for naming variables in Python:
First character must be a letter or an underscore <br>
After the first character, you may use letters, digits, or underscores <br>
Variable name cannot be a Python keyword (reserved word)
 - Examples: and, or, not, True, False (these come later in the course)
 
Variable names are case-sensitive <br>
Variable name should reflect its use and be descriptive (for example, length=25 is preferable to a=25) <br>

Good: name, pig, length, abc123 <br>
Bad: 23name, var.12, #name <br>
Different: name, NAME, Name <br>

Recall from the Flow Charts slides:

In most programming languages, a variable is 
- a memory location that holds a value. In those languages, when you assign a value to a variable, the value is stored in the variable’s memory location.

In Python, a variable is 
- a memory location that holds the address of another memory location. 
- When you assign a value to a Python variable, the variable will hold the address of the memory location that holds the value. 
-That is why, in Python, instead of saying that a variable “holds” a value, we say that a variable “references” a variable.

See the example below.

```python
item_3_price = item_2_price
print ("item_1_price address: ", id(item_1_price),
       "\nitem_2_price address: ", id(item_2_price),
       "\nitem_3_price address: ",id(item_3_price))
# a statement enclosed in parens can be broken without the line continuation character (\)
#\ is the escape character, followed by special character; \n is newline; more at bottom of notebook
# id() function returns an object's unique ID or memory address
```

# Python Math Operators

" + "	Addition ------------ Adds two numbers 

" − "	Subtraction --------- Subtracts one number from another 

" * "	Multiplication ------ Multiplies one number by another 

" / "	Division ------------ Divides one number by another and gives the result as a floating-point number 

" // "	Integer division ---- Divides one number by another and gives the result as a whole number 

" % "	Remainder ----------- Divides one number by another and gives the remainder AKA modulus

" ** "	Exponent ------------ Raises a number to a power

```python
num1 = 5   # An assignment statement
num2 = 10
num3 = -10

result_add = num1+num2    # A math expression performs a calculation and produces a result 
result_sub = num1-num2
result_mult = num1*num2

result_iDiv = num1//num2        # integer division with positive result rounds down
result_iDivneg = num1//num3     # integer division with negative result rounds away from zero
```

 __[Wondering why? See why-pythons-integer-division-floors](http://python-history.blogspot.com/2010/08/why-pythons-integer-division-floors.html)__

```python

result_fDiv = num1/num2
result_exp = num1**num2
result_rem = num1%num2   # known as the modulus operator

print(result_add, result_sub, result_mult, result_iDiv, result_iDivneg, result_fDiv, result_exp, result_rem)
# Look at the printout below.  Are you happy with it?
```

##### Output formatting
Removing endline or adding separators or escape characters <br>
... Achieved via different arguments passed to print() function (end=xx, sep=xx, etc.) <br>

An escape character is preceded with a backslash (\\) inside a string literal. When a string literal that contains escape characters is printed, the escape characters are treated as special commands that are embedded in the string.

Special Commands:

\\n   new line <br>
\\t   tab <br>
\\'   print single quote <br>
\\"   print double quote <br>
\\\\  print backslash <br>




Formatting numbers, inserting comma separators, and specifying minimum field width <br>
... Can be achieved with format() function's specifier (e.g. '10,.2f', '.2e')

<b>Note that Python 3.6 adds the f-string (formatted string literal) to format output; all examples in latest text use the f-string</b>
