<ui> #1: Write a main() function that asks the user to enter the amount of a purchase and whether the item is taxable.  A (Boolean) tax flag is set based on the item's taxable status.
    
The main() function uses the tax flag to determine whether the item is taxable and calculates the state and city tax (if any) on the purchase and then passes the purchase amount and the state and city tax amounts to a function named print_detail which accepts the purchase amount and the state and city taxes as arguments and prints the resulting details about the sale.
    
The state sales tax is 4 percent and the city sales tax is 4.875 percent.
    
print_detail displays the amount of the purchase, the state sales tax, the city sales tax, the total sales tax, and the total of the sale (which is the sum of the amount of purchase plus the total sales tax).
    
The tax amounts are zero for non-taxable purchases (as indicated by the flag).    
    
Use global constants for the state and city tax rates.
    
Use f-strings for formatting the output.
    
Conditionally execute the main() function.    

Hint: Use the value 0.040 to represent 4.0 percent, and 0.04875 to represent 4.875 percent. <br>
    
Sample output follows.    

```python
#ICE6_1 Exer-1 

#Function to ask user for purchase amount and determine if item is taxable
#if the purchase amount is taxable provide the state and city tax rates
def main():
    purchase_amount = float(input("Enter the purchase amount: "))
    taxable = input("Is the item taxable? (Enter y or n): ")
    
    if taxable == 'y':
        STATE_TAX_RATES =  0.040
        CITY_TAX_RATES =  0.04875
        total_tax = STATE_TAX_RATES + CITY_TAX_RATES
        sale_total = purchase_amount * total_tax + purchase_amount
    elif taxable == 'n':
        total_tax = 0.00
        sale_total = purchase_amount    
        CITY_TAX_RATES = 0.00
        STATE_TAX_RATES = 0.00
    else:
        print("invalid input")

    print_detail(purchase_amount, CITY_TAX_RATES, STATE_TAX_RATES, total_tax, sale_total)

    
#A function to print purchase amount, state and city tax, total tax and total sale
def print_detail(purchase_amount, CITY_TAX_RATES, STATE_TAX_RATES, total_tax, sale_total):
    print(f"""\t  Purchase amount:   \t{purchase_amount:,.2f}
          State tax:  \t\t{STATE_TAX_RATES:,.5f}
          City tax:   \t\t{CITY_TAX_RATES:,.5f}
          Total tax:  \t\t{total_tax:,.5f}
          Sale total: \t\t{sale_total:,.2f}""")
    
#Execute main function
main()
```

```python
#ICE6_2 Exer-2
#Global Constant tax rates for city and state level
STATE_TAX_RATES =  0.040
CITY_TAX_RATES =  0.04875
    
#Ask the user for purchase amount
#Determine if it is taxable
#Call calculate function and retuned values
#Print the values and taxes
def main():
    purchase_amount = float(input("Enter the purchase amount: "))
    taxable = input("Is the item taxable? (Enter y or n): ")    
    if taxable == 'y':
        STATE_TAX = STATE_TAX_RATES
        CITY_TAX = CITY_TAX_RATES
    elif taxable == 'n':
        CITY_TAX = 0.00
        STATE_TAX = 0.00
    else:
        print("Invalid input")
    
    total_tax, sale_total = calculate_detail(purchase_amount, STATE_TAX, CITY_TAX,)


    print(f"""\t  Purchase amount:   \t{purchase_amount:,.2f}
          State tax:  \t\t{STATE_TAX:,.5f}
          City tax:   \t\t{CITY_TAX:,.5f}
          Total tax:  \t\t{total_tax:,.5f}
          Sale total: \t\t{sale_total:,.2f}""")


#Calculate the total-tax and sale-total 
#Return those values to the main function
def calculate_detail(purchase_amount, STATE_TAX, CITY_TAX):
    total_tax = STATE_TAX + CITY_TAX
    sale_total = purchase_amount * total_tax + purchase_amount
    return total_tax, sale_total


#Execute main function
main()
```
