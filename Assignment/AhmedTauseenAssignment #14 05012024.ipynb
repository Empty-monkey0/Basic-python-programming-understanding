Program 6-6 Average of Numbers

Assume a file containing a series of integers is named numbers.txt and exists on the computer’s disk. Write a program that calculates the average of all the numbers stored in the file.

```python
# ICE14_1 Programming Exercise 6, Average of Numbers
#Defining main functionto find the average of nuumbers
def main():
    
    #Initialize total sum and count variables
    total = 0.0
    count = 0
    
    # Open the file named 'numbers.txt' in read mode.
    infile = open('numbers.txt', 'r')
    
    # Iterate over each line in the opened file. To find the count and total
    for line in infile:
        val = float(line)
        count += 1
        total += val
        
    #Closes the numbers.txt file
    infile.close()
    
    #Calculates the averagee 
    average = total / count

    #Displays the average value from the .txt file
    print(f"Average: {average}")

#Execute the main function
if __name__ == '__main__':
    main()
```

Program 6-9 Exception Handling

Modify the program that you wrote for __Exercise 6__ so it handles the following exceptions:

•  It should handle any IOError exceptions that are raised when the file is opened and data is read from it.

•  It should handle any ValueError exceptions that are raised when the items that are read 
from the file are converted to a number.

```python
#ICE14_2
#Define the main funtion
def main():
    #Ask user for input on the file name
    file_name = input("Enter file name: ")
    
    try:
        #Opens the file in read mode
        file = open(file_name, 'r')
        #initialize variable to contain a list for the possible numbers in the file
        numbers = []
        
        #Attempting to handle possible exception errors
        try:
            #Iterates over each line in the file
            for line in file:
                try:
                    # Attempt to convert the line to an integer and strips any leading characters
                    number = int(line.strip())
                    #Add the number to the list of numbers
                    numbers.append(number)
                #Catch any non numeric data and display them and close he program
                except ValueError:
                    print(f"Non-numeric data found in the file: {line.strip()}")
                    file.close()
                    return
            #Calulates the average within the list if it is not empty 
            if numbers:
                average = sum(numbers) / len(numbers)
                print(f"Average: {average}")
                
        #closes the file
        finally:
            file.close()
            
    #Catches the error if file cannot be opened or found
    except IOError:
        print("An error occurred while trying to read the file.")

#Execute the main function
if __name__ == '__main__':
    main()
```
