```python
#ICE8_2 Exer-3
#Constant vairable for calories per min lost
CALORIES_PER_MIN = 3.8

#Print the table headings.
print("Minutes\tCalories \n\tBurned")
print("---------------------")

# From 5 to 60 minutes, in steps of 5 minutes
for minutes in range(5, 61, 5):  
    calories_burned = minutes * CALORIES_PER_MIN
    print(f"{minutes:4}\t{calories_burned:5.1f}")
```

<li> Exer-2: Write a program that uses nested loops to collect data and calculate the average rainfall over a period of years. The program should first ask for the number of years. The outer loop will iterate once for each year. The inner loop will iterate twelve times, once for each month. Each iteration of the inner loop will ask the user for the inches of rainfall for that month. When the iterations are complete, the program should display the number of months, the total inches of rainfall, and the average rainfall per month for the entire period.<br>
    
Sample output follows.

```python
#ICE8_4 Exer-2
# Ask for the number of years
years = int(input("Enter the number of years: "))

total_rainfall = 0.0

#Loop through each year and nest a loop for each month
#Ask user for monthly rainfall in each loop
#Calculate total rainfall
for year in range(1, years + 1):
    print(f"\nFor year {year}: ")
    for month in range(1, 13):
        rainfall = float(input(f"Enter the rainfall for month {month}: "))
        total_rainfall += rainfall

    # Calculate total number of months and average rainfall
    total_months = years * 12
    average_rainfall = total_rainfall / total_months

# Display the results
print(f"\nFor {total_months} months")
print(f"Total rainfall: {total_rainfall:.2f} inches")
print(f"Average monthly rainfall: {average_rainfall:.2f} inches")
```
