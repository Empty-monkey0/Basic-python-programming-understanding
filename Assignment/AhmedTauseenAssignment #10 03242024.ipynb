```python
#ICE10_1 Exer-1
# Define function to obtain integer values from user
def get_integers():
    #initialize integer list and continue input
    integers = []
    continue_input = 'y'
    
    # While user select y or Y prompt the user to add intgers
    while continue_input.lower() == 'y':
        value = input("Enter integer value: ")
        
        # Check if the input is an integer
        if value.isdigit():
            integers.append(int(value))
        else:
            print("That wasn't an integer")
        
        continue_input = input("Enter more integer values? (y or n): ")
        
    #return the list values towards the main function
    return integers

# define function to calculate the value list obtained
def calculate(val_list):
    
    # Sort the list in place
    val_list.sort()  
    
    #Calculates sum, avg and determines highest and lowest value
    total = sum(val_list)
    average = total / len(val_list) 
    lowest = val_list[0] 
    highest = val_list[-1] 
    
    # return total, average, lowest, highest, val_list to the main function to be printed
    return total, average, lowest, highest, val_list

#Define main function to obtain values from other functions
#Transfer values obtained to a function to properly order and calculate the values
#Print and display the values calculated 
def main():
    val_list = get_integers()

    total, average, lowest, highest, sorted_values = calculate(val_list)
    
    print(f"Total of the values in the list: {total}")
    print(f"Average of the values in the list: {average}")
    print(f"Lowest value in the list: {lowest}")
    print(f"Highest value in the list: {highest}")
    print("Sorted values in the list:", end=" ")
    
    # for every iteration in the sorted value list print the associated value
    for i in val_list:
        print(i, end=' ') 
    
#Execute the main function    
if __name__ == '__main__':
    main()
```

<ui> #3- Write code to create a two-dimensional list variable called scores that has five columns and 20 rows. Initialize the  values of the list elements to zero using the repetition operator. Then, update the values in the rows and columns by using the randint function in the range of 0-25. <br>

Print the results in the format illustrated below.

```python
#ICE10_1 Exer-3
import random

#Initialize number of rows and coloumns constants
ROWS = 20
COLS = 5

#Initialize the values of the list elements to zero using the repetition operator
scores = [[0] * COLS for i in range(ROWS)] 

#Updating the values in the rows and columns by using 
#the randint function in the range of 0-25
for r in range(ROWS):
    for c in range(COLS):
        scores[r][c] = random.randint(0, 25)

#Printing the scores
for r in scores:
    for value in r:
        print(f"{value:3}", end=" ")
    print()
```
