#1: Write a main() function which prompts the user for an input string and passes it to a function that accepts the string as an argument and returns the number of vowels contained in the string.

The application should have another function that also accepts a string as an argument and returns the number of consonants that the string contains.

The main() function should display the number of vowels and the number of consonants contained in the string, using the results returned from the function calls.

Keep in mind that the string might contain spaces separating words, that there may be special characters in the string and that the letters might be a mix of upper and lower case.

Sample output follows.    

```python
#ICE9_1 Exer-1
#Define a function that counts vowels of the arg
def count_vowels(input_string):
    
    #local varible containing all vowels
    vowels = "aeiouAEIOU"
    
    #Accumulator variable to keep count
    vowel_count = 0
    #A loop going through every character in the input
    for char in input_string:
        #Compares input character to see if it matched vowels
        if char in vowels:
            #When true keeps track and adds 1 to accumulator variable
            vowel_count += 1
            
    # return vowel count to the main function
    return vowel_count

#Define function that counts cononates of the arg 
def count_consonants(input_string):
    
    #local varible containing all consonants
    consonants = "bcdfghjklmnqrstvwxyxBCDFGHJKLMNPQRSTVWXYZ"
    
    #Accumulator variable to keep count is initialized
    consonant_count = 0
    
    #For loop going through every character in the input
    for char in input_string:
        #Compares input character to see if it matched consonants
        if char in consonants:
            #When true keeps track and adds 1 to accumulator variable
            consonant_count += 1
    #return consonant count to the main function
    return consonant_count


#Define the main function to ask user input and print a string
def main():
    #Ask user for input string
    user_input = input("Enter a string: ")
    
    #Call the function and store the return values for vowels and consonants
    num_vowels = count_vowels(user_input)
    num_consonants = count_consonants(user_input)
    
    #Displays string of text showing the vowels and consonants
    print(f' "{user_input}" contains {num_vowels} vowels and {num_consonants} consonants.')
    
#Execute the main function
if __name__ == '__main__':
    main()
    
```

<ui> #2: Write a main() function which prompts the user for an input string and passes it to a function that accepts the string as an argument and counts and returns the number of uppercase, lowercase, numbers and spaces contained in the string.
    
Call this function from main(), which prints out the results returned from the function.    
    
Sample output follows.

```python
#ICE9_1 Exer-2
#Define the main function to ask user for input
#Call the analyze string function of the return values to
#Display the results of lower,upper,digit,spaces char amount
def main():
    
    input_string = input("Enter a string: ")
    
    uppercase, lowercase, digits, spaces = analyze_string(input_string)

    print(f'"{input_string}" contains:')
    print(f"{uppercase} upper case characters")
    print(f"{lowercase} lower case characters")
    print(f"{digits} digits")
    print(f"{spaces} spaces")


#define analyze string function that counts the 
#Uppercase, lowercase, digits and spaces in the provided string
def analyze_string(input_string):
    
    #initiliaze accumulator variables
    uppercase_count = 0
    lowercase_count = 0
    digit_count = 0
    spaces_count = 0
    
    #A loop that goes through each value in a string
    #To check if character is upper, lower, digit, space
    #Adds count in the corresponding accoumulator variable
    for char in input_string:
        if char.isupper():
            uppercase_count += 1
        elif char.islower():
            lowercase_count += 1
        elif char.isdigit():
            digit_count += 1
        elif char.isspace():
            spaces_count += 1

    #Returns the counts of each corresponding value to the main function        
    return uppercase_count, lowercase_count, digit_count, spaces_count

#Execute main function
if __name__ == "__main__":
    main()
```

<ui> #3: Write a main() function which prompts the user for an input string and passes it to a function that returns the character that appears most frequently in the string, along with the number of times this character appeared.  Don't count a space as a character.
    
__Treat upper and lower case versions of a character as the same character, as in the sample output, below.__
    
main() print out the results returned from the function.<br>
    
Sample output follows.

```python
#ICE9_1 Exer-3
#Define main fauntion to ask user for a string input
#Call frequent char function to find the most freqeuent character and how many time
#Display that within a print function
def main():
    input_string = input("Enter a string: ")
    most_frequent_char, frequency = frequent_char(input_string)
    print(f'The character that appears most frequently is "{most_frequent_char}" and it appears {frequency} times.')

#Define freqent character function
def frequent_char(input_string):
    
    # Initialize lists to keep track of characters and their counts
    characters = []
    counts = []
    
    # Convert the string to lowercase and remove spaces
    format_string = input_string.lower().replace(" ", "")
    
    # Loop through each character in the formatted string
    #To determine if the character is already in the list and increase its count
    for char in format_string:
        if char in characters:
            index = characters.index(char)
            counts[index] += 1
        else:
            # If it's a new character, add it to the list and add its count
            characters.append(char)
            counts.append(1)
    
    # Go through counts list find the max value
    max_count = max(counts)
    
    #Find the index associated to the max char which has the max count
    max_index = counts.index(max_count)
    
    #Goes to character list wih the index associated with the most frequent character
    most_frequent_char = characters[max_index]
    
    # Return the most frequent character converted to uppercase and its count
    return most_frequent_char.upper(), max_count

#Execute main function
if __name__ == '__main__':
    main()
```
