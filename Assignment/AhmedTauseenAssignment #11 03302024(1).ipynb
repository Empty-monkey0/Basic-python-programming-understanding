```python
#ICE11 Question #3
# Given dictionaries
d1 = {1:10, 2:20}
d2 = {1:20, 3:30, 4:40}
d3 = {5:50, 6:60, 3:60}

# Create a new dictionary by concatenating the given dictionaries
combined_dictionary = {}

# Loop over the list of dictionaries
for dictionary in [d1, d2, d3]:  
    #loop over each dictionary's items and add them over to a seperate dictionary
    for key, value in dictionary.items():  
#Determine if the a key value repeats and to add that value if it does
        if key in combined_dictionary:
            combined_dictionary[key] += value 
        else:
            combined_dictionary[key] = value  
            
#Display the new dictionary
print(combined_dictionary)
```

<ui> #4: - Write a function that takes a word as a parameter and __returns a dictionary__ that tells us how many times each letter appears in the word (no need to account for absent letters).
        
Display your output in the main() function.
    
__Develop your logic using dictionaries only.__    
    
Do not count any character which is not a letter; for each letter, there is a combined count of lowercase and upper case versions of that letter; do not count "a" and "A" as separate characers.
    
__Note that the output does not display the brackets that Python uses to denote the contents of a dictionary;__ your output, as in the sample output, should show that you can control the formatting __by not displaying brackets,__ unlike the problem above.

Sample output follows.

```python
#ICE11 Question #4
#Define Function to count the letters provided in the user input
def count_letters(user_input):
    #initialize empty dictionary
    letter_counts = {}
    
    # Convert the word to lowercase for uniform counting
    #Go through and add a dictionary value For every char in the user input
    for char in user_input.lower():  
        # Check if the character is a letter
        if char.isalpha():
            #Determine whether the char is new or needs to be added to an old char
            if char in letter_counts:
                letter_counts[char] += 1
            else:
                letter_counts[char] = 1
                
    #Return dictionary to the main function
    return letter_counts

#Define main function to prompt user for input
#Pass function data to count_letters and gain the dictionary
#print and display the dictionary values with brackets
def main():
    user_input = input("Enter a word: ")
    letter_counts = count_letters(user_input)
    print(str(letter_counts).replace("{","").replace("}", "").replace("'","").replace(",",""))

#Execute the main function
if __name__ == '__main__':
    main()
```

<ui> #5: Write a program using functions that keeps names and email addresses in a dictionary as key-value pairs.
    
The program should have a function that displays a menu (see Program 9-2 for an example of a menu) that lets the user choose from the following options, each of which is a separate function:

   - look up and return a person’s email address if it exists,
   - add a new name and email address,
   - change an existing email address and
   - delete an existing name and email address
    
The menu function returns the menu choice to main().    
   
Each function (except main()) gathers any information it needs from the user and accesses the dictionary and prints the results.
    
After executing any of the above functions, control should return to the main() function which uses a while loop (do not use while True) to determine whether the menu should be displayed again, and if so, invokes the appropriate function.
    
The main() function prints the messge "Thank you for using the email system" when the user ends the session.    
    
Sample output follows.    

```python
#ICE11 Question #5
#This program uses a dictionary to keep
# names and emails.

# Global constants for menu choices
LOOK_UP = 'l'
ADD = 'a'
UPDATE = 'u'
DELETE = 'd'
QUIT = 'q'

# main function
def main():
    # Create an empty dictionary.
    emails = {}

    # Initialize a variable for the user's choice.
    choice = 0

    while choice != QUIT:
        # Get the user's menu choice.
        choice = get_menu_choice()

        # Process the choice.
        if choice == LOOK_UP:
            look_up(emails)
        elif choice == ADD:
            add(emails)
        elif choice == UPDATE:
            update(emails)
        elif choice == DELETE:
            delete(emails)
        else:
            print("Thank you for using the email system")

# The get_menu_choice function displays the menu
# and gets a validated choice from the user.
def get_menu_choice():
    print()
    print('Email Address Book')
    print('---------------------------')
    print('\nLook up email address (1)')
    print('Add a new name and email (a)')
    print('Updating an existing email (u)')
    print('Delete an existing email (d)')
    print('Quit (q)')
    print()

    # Get the user's choice.
    choice = str(input('Enter what you wish to do: '))

    # Validate the choice.
    while choice != QUIT and choice != LOOK_UP and choice != ADD and choice != UPDATE and choice != DELETE :
        choice = str(input('Enter one of above options.\nEnter what you wish to do: '))

    # return the user's choice.
    return choice

# The look_up function looks up a name in the
# emails dictionary.
def look_up(emails):
    # Get a name to look up.
    name = input('Enter a name: ')

    # Look it up in the dictionary and provide formatting
    print(f"Name: {name} Email: {emails.get(name, 'None')}")

# The add function adds a new entry into the
# emails dictionary.
def add(emails):
    # Get a name and email.
    name = input('Enter a name: ')

    # If the name does not exist, add it.
    if name not in emails:
        email = input('Enter a email: ')
        emails[name] = email
        print(f"{name} added")        
    else:
        print('That entry already exists.')

# The UPDATE function UPDATEs an existing
# entry in the emails dictionary.
def update(emails):
    # Get a name to look up.
    name = input('Enter a name: ')

    if name in emails:
        # Get a new email.
        email = input('Enter the new email: ')
        print(f"{name} updated")

        # Update the entry.
        emails[name] = email
    else:
        print(f'{name} name not found.')

# The delete function deletes an entry from the
# emails dictionary.
def delete(emails):
    # Get a name to look up.
    name = input('Enter a name: ')

    # If the name is found, delete the entry.
    if name in emails:
        del emails[name]
        print(F"{name} deleted")
    else:
        print('That name is not found.')

# Call the main function.
if __name__ == '__main__':
    main()
```
