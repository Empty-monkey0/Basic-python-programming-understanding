### Exer-2: 

An aircraft burns 10,000 pounds of fuel per hour.

Write a Python program that asks the user for the number of miles flown and the duration (in hours) of the flight. Calculate the aircraft’s average speed in miles per hour and number of pounds of fuel consumed and display the results.  __Use the Python convention for a constant in your solution.__<br><br> 
Create the pseudocode or flowchart first and then write the corresponding code.<br><br>
Sample output follows.

```python
#Exer-2 Solution
# pounds of fuel burned per hour Constant
BURN_RATE = 10000

#asks the user for the number of miles flown in hours
miles_flown = float(input("Enter the number of miles flown: "))

# ask the user for the duration (in hours) of the flight
duration = float(input("Enter the duration of the flight in hours: "))

#Calculate average speed
average_speed = miles_flown / duration

#Calculate fuel burned
fuel_consumed = duration * BURN_RATE

#Display aircraft’s average speed and pounds of fuel consumed
print(f'The aircraft’s average speed is {average_speed} miles per hour.')
print(f'The aircraft consumed {fuel_consumed} pounds of fuel.')
```

### Exer-3: 

Many people keep time using a 24-hour clock (11 is 11 AM and 23 is 11 PM, 0 is midnight). As an example, if it is currently 13 (1 PM) and you set your alarm to go off in 50 hours, it will go off at 15 (3 PM) two days in the future.  Ask the user for the time now (in hours), and then ask for the number of hours to wait for the alarm. Write the pseudocode or flowchart first and the corresponding Python program to solve the general version of the above problem. Your program should output what the time will be on the clock in 24-hour format when the alarm goes off. Consider only whole hours, i.e., do not worry about minutes and seconds for now.

You should test various scenarios to make sure that your code performs correctly in all cases.

For example, test the code:
-   with a current time before midnight and the alarm going off before midnight. <br> 
-   with a current time before midnight and the alarm going off after midnight.
-   with the alarm set for more than 24 hours in the future.

You may wish to refer to Program 2-17 in the text for an example of the use of modulo arithmetic, which __must be used to solve this problem__.

 __Use the Python convention for a constant (the number of hours in a day) in your solution.__<br>

Sample output follows; __remember that you must demonstrate the successful execution of the code in different scenarios, if applicable, by displaying multiple output cell(s).__

```python
#Exer-3 Solution
#hours in a single day constant
DAY_RATE = 24

#Ask the user current military time and 
current_Mtime = float(input("Enter the current military time (hours only): "))
timer_hours = float(input("Enter the number of hours until the alarm sounds: "))

#calculation when the alarm will go off 
alarm_time = int((current_Mtime + timer_hours) % DAY_RATE)

#Display that that the alarm will go off 
print(f"The alarm will go off at: {alarm_time} hours")
```

### Exer-4: 

Body Mass Index (BMI) can be a helpful health measure for large groups of people, but it won’t tell you much about yourself (__[Is BMI a Scam?](https://www.nytimes.com/2021/05/18/style/is-bmi-a-scam.html))__

Write a program that asks the user for their weight (in pounds) and height (in inches). Calculate and output the BMI to two decimal places.

Formula for BMI: 703 * weight (lbs) / [height (in)]^2

__Use the Python convention for a constant in your solution.__ 

Sample output follows.

```python
#Exer-4 Solution
#Number used to convert pounds to kilograms and inches to meters
CONVERSION = 703

#asks the user for weight and height
weight = float(input("Enter weight (in pounds): " ))
height = float(input("Enter height (in inches): "))

# BMI calculation
BMI = CONVERSION * weight / ((height)**2)

#Display BMI value
print(f'Your BMI is {BMI:.2f}')
```
