<li> Exer-3: Write a program that asks the user to enter the amount that he or she has budgeted for a month. A loop should then prompt the user to enter each of his or her expenses for the month and keep a running total. When the loop finishes, the program should display the amount that the user is over or under budget.<br>

__Use a WHILE loop which tests a sentinel for a value of 0__ to end the expenses loop and __use an augmented assignment operator__ to maintain the running total of expenses.
    
Sample output follows.

```python
#ICE7_2 Exer-3
def main():
    #Ask user for the months budget 
    budget = float(input("Enter amount budgeted for the month: "))
    
    #initialized variables
    total_expenses = 0.0
    expense = 0.001
    
    #While loop ends when 0 is entered and determines total expense
    while expense > 0:
        expense = float(input("Enter an amount spent (0 to quit): "))
        total_expenses += expense
    
    #Displays the budget and expenses
    print(f"Budgeted: ${budget:,.2f}")
    print(f"Spent: ${total_expenses:,.2f}")
    
    
    #Determine if expenses are over or under budget and calculate the values
    if total_expenses > budget:
        over_budget = total_expenses - budget
        print(f"You are ${over_budget:,.2f} over budget. PLAN BETTER NEXT TIME!")
    elif total_expenses < budget:
        under_budget = budget - total_expenses
        print(f"You are ${under_budget:,.2f} under budget. Good job staying on track!")
    else:
        print("You spent exactly your budget.")

#execute main() function
if __name__ == '__main__':
    main()
```
