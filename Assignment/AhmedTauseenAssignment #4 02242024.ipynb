Sample output follows.    

```python
#ICE4_3 Exer-3 Solution

#Constants for Overtime and Hour limit
REGULAR_HOURS_LIMIT = 40
DOUBLE_TIME_HOUR_LIMIT = 60
OVERTIME_FACTOR = 1.5
OVERTIME_FACTOR2 = 2.0

#Ask user for hours worked and hourly wage
hours_worked = float(input('Enter hours worked: '))
hourly_wage = float(input('Enter dollars paid per hour: ')) 

#formula for overtime hours worked
overtime_hours_worked = hours_worked - REGULAR_HOURS_LIMIT 
doubletime_hours_worked = hours_worked - DOUBLE_TIME_HOUR_LIMIT

#Determine if hours worked surpass regular hour limit and if double time applies
if hours_worked > DOUBLE_TIME_HOUR_LIMIT:
    regular_hours_worked = REGULAR_HOURS_LIMIT
    overtime_hours_worked = DOUBLE_TIME_HOUR_LIMIT - REGULAR_HOURS_LIMIT
elif hours_worked > REGULAR_HOURS_LIMIT:   
    regular_hours_worked = REGULAR_HOURS_LIMIT
    doubletime_hours_worked = 0
else:
    doubletime_hours_worked = 0
    overtime_hours_worked = 0 
    regular_hours_worked = hours_worked 

#Calulate wages under regular hours
normal_wage = (regular_hours_worked * hourly_wage)

#Calculate wages under overtime
overtime_wage = ((OVERTIME_FACTOR*hourly_wage)*overtime_hours_worked)

#Calculate wages during doubletime
double_time_wage = ((OVERTIME_FACTOR2*hourly_wage)* doubletime_hours_worked)

#Calculate total wages
total_wages = normal_wage + overtime_wage + double_time_wage


#Display hours worked, hourly wage and total wage
print(f'Wages for {hours_worked} hours at ${hourly_wage:,.2f} per hour are ${total_wages:,.2f}')
```

<li> Exer-5: Write an if-elif-else statement using Boolean expressions and the Gregorian approach to determine whether a given year is a leap year. Here are the conditions: <br><br>
    Determine whether the year is divisible by 100. If it is, then it is a leap year if and only if it is also divisible by 400. For example, 2000 is a leap year, but 2100 is not. <br><br>
    If the year is not divisible by 100, then it is a leap year if and only if it is divisible by 4. For example, 2008 is a leap year, but 2009 is not.
    
__Your solution must use the modulus operator to determine whether a year is divisible by 100, 400 etc.__    
    
Sample output follows.    

```python
#4_3 Exer-5 Solution

#Ask user for the year
year = int(input("Enter a year: "))

#Determine id the year is a leap or not
if year%100==0 and year%400 == 0:
	print(f"{year} is a leap year")
elif year%100 != 0 and year%4 == 0:
	print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
```
