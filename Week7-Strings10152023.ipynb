# String Slicing: Creating substrings from a string with indexing
### General format is string[start : end : step]
An index and a slice have different behaviors (semantics).
Degenerate slice indices are handled gracefully: an index that is too large is replaced by the string size, an upper bound smaller than the lower bound returns an empty string.  See https://docs.python.org/3.0/tutorial/introduction.html

```python
sub_t = my_text[]    # can't completely omit slice expression
```

```python
# What will be the output in each of the cases below?
sub_t = my_text[:]   # default value of start index is 0; default value of end index is len(string)
sub_t
```

```python
sub_t = my_text[0]
sub_t
```

```python
sub_t = my_text[1:]
sub_t
```

```python
sub_t = my_text[:2]
sub_t
```

```python
sub_t = my_text[1:8]
sub_t
```

```python
sub_t = my_text[::2]     # Step of 2
sub_t
```

```python
sub_t = my_text[1:8:3]
sub_t
```

```python
sub_t = my_text[1:100:3]  # If end index specifies position beyond end of string, uses len(string) 
sub_t
```

```python
sub_t = my_text[10:8:3]   # upper bound (8) less than lower bound (10) 
sub_t     # upper bound (end) smaller than the lower bound (start) returns an empty string
```

#### Fun facts about slicing indexes???? 
Invalid indexes do not cause slicing expressions to raise an exception. For example:
- If the end index specifies a position beyond the end of the string, Python will use the length of the string instead.
- If the start index specifies a position before the beginning of the string, Python will use 0 instead.
- If the start index is greater than the end index, the slicing expression will return an empty string.

## String Built-in Methods
- <b> Testing </b> whether a string has certain features such as numbers, characters, both, upper/lower case, spaces, etc.

```python
my_str = "              i registered for CIS3120 in Spring 2022!"
# Check if the string contains each of upper/lower case characters, numbers, and spaces?
# Methods to use: isupper, islower, isalnum, isalpha, isdigit, isspace

for ch in my_str:
    if ch.isupper():
        up_flag = True
    if ch.islower():
        low_flag = True
    if ch.isdigit():
        digit_flag = True
    if ch.isspace():
        space_flag = True

if up_flag:
    print("Has an uppercase letter")
if low_flag:
    print("Has a lowercase letter")
if digit_flag:
    print("Has a number")
if space_flag:
    print("Contains spaces")
```
