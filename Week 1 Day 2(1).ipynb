Enter Jupyter command mode by pressing Esc or using the mouse to click outside a cell’s editor area.

#### To toggle line numbering in a single cell, click  outside a cell’s editor area and press Esc (top-left corner of keyboard) l (lowercase L).

#### To toggle line numbering in the notebook as a whole, click  outside a cell’s editor area and press shift-l (lowercase L).

To view all shortcuts, bring up the command palette: Ctrl-Shift+P (Cmd+Shift+P on Mac).
